```python
import pandas as pd
lista_nombres=["alexander","alejandra","dario","geovany","ana","santiago","rita","jaime","fernando","rodrigo","mentor","ejecutor"]
lista_pesos=[69,57,64,65,62,65,90,68,72,65,74,78]#peso en kg
lista_alturas=[1.78,1.60,1.64,1.70,1.57,1.62,1.63,1.65,1.72,1.68,1.71,1.78] #altura en m

dicc_datos={"nombres":lista_nombres,"pesos":lista_pesos,"alturas":lista_alturas}
df=pd.DataFrame(dicc_datos) #convierte al diccionario en un dataframe(marco de datos o tabla o dataset)
df["imc"] = df["pesos"] / (df["alturas"]**2)
def clasificacion(imc):
    if imc < 18.5:
        return "bajo peso"
    elif imc< 25:
        return "peso normal"
    else:
      return "sobrepeso"
df["clasificacion"] = df["imc"].apply(clasificacion)
df
```
