# Task
Perform a logistic regression (classification) with classification metrics (confusion matrix and classification report) including an error histogram.

## Preparación de los datos

### Subtask:
Seleccionar las variables relevantes para la regresión logística, codificar la variable objetivo si es necesario, y dividir los datos en conjuntos de entrenamiento y prueba.


**Reasoning**:
Identify relevant features, check and encode the target variable, define X and y, and split the data for logistic regression.



```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Identify relevant features. Including both numerical and categorical features for a more comprehensive model.
# We need to consider how to handle categorical features later (e.g., one-hot encoding).
feature_cols = ['Year', 'Engine_Size_L', 'Mileage_KM', 'Price_USD', 'Region', 'Color', 'Fuel_Type', 'Transmission']
X = df[feature_cols]
y = df['Sales_Classification']

# Check unique values of the target variable
print("Unique values of Sales_Classification:", y.unique())

# Encode the target variable 'Sales_Classification' as it's categorical
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.25, random_state=42)

print("\nShape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

## Entrenamiento del modelo

### Subtask:
Entrenar un modelo de regresión logística con los datos de entrenamiento.


**Reasoning**:
Import the LogisticRegression class and train the model using the training data.



```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

**Reasoning**:
The error indicates that the model cannot handle the categorical features in their current string format. These categorical features need to be encoded into a numerical format before training the model. One-hot encoding is a suitable method for this.



```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Identify categorical columns from the original feature list
categorical_features = ['Region', 'Color', 'Fuel_Type', 'Transmission']
numerical_features = ['Year', 'Engine_Size_L', 'Mileage_KM', 'Price_USD']

# Create a column transformer to apply one-hot encoding to categorical features
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# Create a pipeline that first preprocesses the data and then trains the logistic regression model
model = Pipeline(steps=[('preprocessor', preprocessor),
                        ('classifier', LogisticRegression())])

# Train the model using the training data
model.fit(X_train, y_train)
```

## Predicción

### Subtask:
Realizar predicciones sobre el conjunto de prueba utilizando el modelo entrenado.


**Reasoning**:
Utilize the trained logistic regression model to make predictions on the test dataset and store these predictions in a new variable.



```python
y_pred = model.predict(X_test)
```

## Evaluación del modelo

### Subtask:
Calcular y mostrar la matriz de confusión y el reporte de clasificación para evaluar el rendimiento del modelo de clasificación.


**Reasoning**:
Calculate and print the confusion matrix and classification report to evaluate the classification model's performance.



```python
from sklearn.metrics import confusion_matrix, classification_report

# Calculate the confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)

# Print the confusion matrix
print("Matriz de Confusión:")
display(conf_matrix)

# Define target names based on the LabelEncoder classes
# Make sure to use the classes_ attribute of the fitted LabelEncoder
target_names = le.classes_

# Calculate the classification report
class_report = classification_report(y_test, y_pred, target_names=target_names)

# Print the classification report
print("\nReporte de Clasificación:")
display(class_report)
```

## Análisis de errores

### Subtask:
Calcular los errores de predicción (diferencia entre las etiquetas reales y predichas) y visualizar su distribución con un histograma.


**Reasoning**:
Calculate the prediction errors and plot their distribution as a histogram.



```python
import matplotlib.pyplot as plt

# Calculate the prediction errors
prediction_errors = y_test - y_pred

# Create a histogram of the prediction errors
plt.figure(figsize=(10, 6))
plt.hist(prediction_errors, bins=50)
plt.title('Histograma de Errores de Predicción')
plt.xlabel('Error de Predicción')
plt.ylabel('Frecuencia')
plt.show()
```
