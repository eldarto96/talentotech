```python
def registrar_ventas():
    producto=input("Ingrese el nombre del producto: ")
    cantidad=int(input("Ingrese la cantidad vendida: "))
    precio=float(input("Ingrese el precio unitario: "))
    total=cantidad*precio

    return{"producto":producto,
           "cantidad":cantidad,
           "precio":precio,
           "total":total}
```

```python
def mostrar_menu():

  print("-----Menu-----")

  print("1. Registrar venta")

  print("2. Mostrar ventas")

  print("3. Analitica de Ventas")

  print("4. Salir")

mostrar_menu()
```

```python
import pandas as p

while True:
    mostrar_menu()
    opcion = input("Ingrese su opcion: ")

    if opcion == "1":
        cantidad_registros = int(input("Ingrese la cantidad de registros a registrar: "))
        for i in range(cantidad_registros):
            venta = registrar_ventas()
            ventas.append(venta)
            print("Venta registrada con exito")
    elif opcion == "2":
        if len(ventas) == 0:
            print("No hay ventas registradas")
        else:
            print("-----Ventas-----")
            for venta in ventas:
                print(f"Producto: {venta['producto']}, Cantidad: {venta['cantidad']}, Precio Unitario: {venta['precio']}, Total: {venta['total']}")
    elif opcion == "3":
        if len(ventas) == 0:
            print("No hay ventas registradas")
        else:
            df = p.DataFrame(ventas)
            print(df)
    elif opcion == "4":
        print("Gracias por usar el programa")
        break
    else:
        print("Opcion invalida")
```
