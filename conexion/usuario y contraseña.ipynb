```python
def pedir_entero(mensaje):
    """Pide un número entero positivo."""
    while True:
        try:
            valor = int(input(mensaje))
            if valor >= 0:
                return valor
            else:
                print("⚠ Debe ser un número positivo.")
        except ValueError:
            print("⚠ Debes ingresar un número válido.")


def pedir_flotante(mensaje, minimo=None, maximo=None):
    """Pide un número decimal con rango opcional."""
    while True:
        try:
            valor = float(input(mensaje))
            if (minimo is None or valor >= minimo) and (maximo is None or valor <= maximo):
                return valor
            else:
                print(f"⚠ El valor debe estar entre {minimo} y {maximo}.")
        except ValueError:
            print("⚠ Debes ingresar un número válido.")


# ============================
# Credenciales
# ============================
credenciales = ("admin", "1234")


# ============================
# Inicio de sesión
# ============================
while True:
    usuario = input("Usuario: ")
    contrasena = input("Contraseña: ")


    if (usuario, contrasena) == credenciales:
        print("\n✅ Inicio de sesión exitoso\n")


        # ============================
        # Variables principales
        # ============================
        estudiantes = []
        correos = set()
        ubicacion_sede = None


        # ============================
        # Menú principal
        # ============================
        while True:
            print("\n--- MENÚ PRINCIPAL ---")
            print("1. Registrar estudiantes")
            print("2. Registrar ubicación de la sede")
            print("3. Registrar correos")
            print("4. Registrar datos detallados de un estudiante")
            print("5. Salir")


            opcion = input("Selecciona una opción: ")


            if opcion == "1":
                cantidad = pedir_entero("¿Cuántos estudiantes quieres registrar? ")
                for _ in range(cantidad):
                    nombre = input("Nombre del estudiante: ")
                    estudiantes.append(nombre)
                print("📋 Lista de estudiantes:", estudiantes)


            elif opcion == "2":
                latitud = pedir_flotante("Ingresa la latitud: ")
                longitud = pedir_flotante("Ingresa la longitud: ")
                ubicacion_sede = (latitud, longitud)
                print("📍 Ubicación de la sede:", ubicacion_sede)


            elif opcion == "3":
                cantidad = pedir_entero("¿Cuántos correos registrarás? ")
                for _ in range(cantidad):
                    email = input("Correo: ")
                    correos.add(email)
                print("📧 Correos únicos registrados:", correos)


            elif opcion == "4":
                estudiante = {
                    "nombre": input("Nombre: "),
                    "edad": pedir_entero("Edad: "),
                    "carrera": input("Carrera: "),
                    "notas": []
                }


                cantidad_notas = pedir_entero("¿Cuántas notas vas a ingresar? ")
                for _ in range(cantidad_notas):
                    nota = pedir_flotante("Nota: ", minimo=0, maximo=5)
                    estudiante["notas"].append(nota)


                estudiante["promedio"] = sum(estudiante["notas"]) / len(estudiante["notas"])
                print("📊 Datos del estudiante:", estudiante)


            elif opcion == "5":
                print("👋 Saliendo del programa...")
                break


            else:
                print("⚠ Opción no válida.")


        break  # Rompe el bucle de login después de salir del menú


    else:
        print("❌ Usuario o contraseña incorrectos.")
        reintentar = input("¿Intentar de nuevo? (s/n): ").lower()
        if reintentar != "s":
            print("👋 Saliendo del programa...")
            break
```
