```python
import pandas as pd

listax=[1,2,2,3,4,4,5,6]
listay=[2,3,4,4,4,6,5,7]

dicc_datos={
    "x": listax,
    "y": listay
}

df=pd.DataFrame(dicc_datos)

df["xy"]=df["x"]*df["y"]
df["x^2"]=df["x"]**2

suma = df.sum()
df.loc["Suma"] = suma

# Número de datos
n = len(listax)

# Obtener valores de las sumas
sum_x = df.loc["Suma","x"]
sum_y = df.loc["Suma","y"]
sum_xy = df.loc["Suma","xy"]
sum_x2 = df.loc["Suma","x^2"]

m = (n*sum_xy - sum_x*sum_y) / (n*sum_x2 - sum_x**2)
b = (sum_y/n) - m*(sum_x/n)

print("Pendiente (m):", round(m,5))
print("Ordenada al origen (b):", round(b,5))

df
```

```python
df=df.drop(index="Suma")
df
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Crear el gráfico de dispersión
plt.scatter(df["x"], df["y"], label='Datos')

# Generar puntos para la recta de regresión
x_reg = np.linspace(df["x"].min(), df["x"].max(), 100)
y_reg = m * x_reg + b

# Graficar la recta de regresión
plt.plot(x_reg, y_reg, color='red', label=f'Recta de Regresión: y = {m:.5f}x + {b:.5f}')

# Añadir etiquetas y título
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gráfico de Dispersión y Recta de Regresión Lineal")
plt.legend()
plt.grid(True)
plt.show()
```
