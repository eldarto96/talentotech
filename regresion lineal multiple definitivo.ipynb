```python
import pandas as pd
import numpy as np
listay=[78.5, 74.3, 104.3, 87.6, 95.9, 109.2, 102.7, 72.5, 93.1, 115.9, 83.8, 113.3, 109.4]
listax1=[7, 1, 11, 11, 7, 11, 3, 1, 2, 21, 1, 11, 10]
listax2=[26, 29, 56, 31, 52, 55, 71, 31, 54, 47, 40, 66, 68]
listax3=[ 6, 15, 8, 8, 6, 9, 17, 22, 18, 4, 23, 9, 8]
listax4=[60, 52, 20, 47, 33, 22, 6, 44, 22, 26, 34, 12, 12]

dicc_datos={"y":listay,"x1":listax1,"x2":listax2,"x3":listax3,"x4":listax4}

df=pd.DataFrame(dicc_datos)

print("orginal dataframe:")
display(df)
# replace the "y" column values  with ones
df["y"]=1

print("\nDataFrame with 'y' column converted to ones:")
display(df)

#convert the modified frame to a NumPy array
matrix_ones_y=df.values
print("\nMatrix with 'y' column converted to ones:")
print(matrix_ones_y)
```

```python
matrix_ones_y
```

```python
matrix_ones_y_transpose=matrix_ones_y.T
print(matrix_ones_y_transpose)
```

```python
res = matrix_ones_y_transpose @ matrix_ones_y
res
```

```python
res_inv = np.linalg.inv(res)
res_inv
```

```python
print(res_inv)
```

```python
betas = res_inv @ matrix_ones_y_transpose @ np.array(listay)
betas
```

```python
X=matrix_ones_y
X
```

```python
ypred = X @ betas
ypred
```

```python
df["y-ypred"]=df["y"]-df["ypred"]
df
```

```python
error_cuadratico_total=df["error^2"].sum()
error_cuadratico_total
```

```python
error_cuadratico_medio=error_cuadratico_total/len(ypred)
error_cuadratico_medio
```
